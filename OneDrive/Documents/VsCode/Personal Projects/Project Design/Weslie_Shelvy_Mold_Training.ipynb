**Run order:** top to bottom. Set Runtime → Change runtime type → **GPU (T4)** first.

**Output:** a `best.onnx` at 640 to drop onto the Pi (`--model best.onnx --size 640`).

````python
# --- OPTION A: Roboflow download ---
from roboflow import Roboflow
rf = Roboflow(api_key="PUT_YOUR_ROBOFLOW_API_KEY")
project = rf.workspace("PUT_WORKSPACE").project("PUT_PROJECT")
version = project.version(1)                 # change version number
dataset = version.download("yolov8")         # segmentation export = yolov8
DATA_YAML = dataset.location + "/data.yaml"
print("data.yaml at:", DATA_YAML)
````

## 4) Check data.yaml
Classes must be **bread** and **mold**. Fix the file if the names/paths are wrong.

````python
import yaml
with open(DATA_YAML) as f: cfg = yaml.safe_load(f)
print(cfg)
print('\nclasses:', cfg.get('names'))   # expect ['bread','mold'] (order matters)
````

## 5) Train (YOLOv8-seg)
- `yolov8s-seg` = good accuracy (use `yolov8n-seg` if training is slow / Pi too slow)
- `imgsz=640` MUST match the Pi (`--size 640`)
- gentle color aug for the NoIR camera; keep flips/rotation/brightness

````python
from ultralytics import YOLO
model = YOLO('yolov8s-seg.pt')   # pretrained backbone, we fine-tune on our data
results = model.train(
    data=DATA_YAML,
    epochs=150,
    patience=30,
    imgsz=640,
    batch=16,
    device=0,
    project='shelvy', name='seg_v1',
    # augmentation (NoIR-friendly)
    hsv_h=0.010, hsv_s=0.30, hsv_v=0.40,
    fliplr=0.5, flipud=0.0, degrees=10, translate=0.1, scale=0.4,
    mosaic=1.0, close_mosaic=10
)
print('best weights:', results.save_dir)
````

## 6) Validate — get the confusion matrix (for TC3)
This runs on your held-out val/test split and writes the confusion matrix + PR curves.

````python
best = YOLO('shelvy/seg_v1/weights/best.pt')
metrics = best.val(data=DATA_YAML, imgsz=640, split='val')
print('mAP50:', metrics.box.map50, ' mAP50-95:', metrics.box.map)
# confusion matrix image is saved in the val run dir:
from IPython.display import Image, display
import glob
for p in glob.glob('shelvy/seg_v1*/confusion_matrix*.png'):
    display(Image(p))
````

## 7) Quick test — predict on a sample (multiple / mixed breads)
Lower `conf` (0.15) matches what we run on the Pi — catches more mold.

````python
import glob
test_imgs = glob.glob(dataset.location + '/valid/images/*.jpg')[:3]  # or your own paths
pred = best.predict(test_imgs, imgsz=640, conf=0.15, save=True)
for r in pred:
    names = [best.names[int(c)] for c in r.boxes.cls] if r.boxes is not None else []
    print(r.path, '->', names)
for p in glob.glob('runs/segment/predict*/*.jpg')[:3]:
    display(Image(p))
````

## 8) Export to ONNX (for the Raspberry Pi)
`opset=12`, `imgsz=640` — matches `detect_pi.py`.

````python
onnx_path = best.export(format='onnx', imgsz=640, opset=12, simplify=True)
print('ONNX saved:', onnx_path)
````

````python
from google.colab import files
files.download(onnx_path)   # downloads best.onnx to your PC
````

## 9) Deploy on the Pi
1. Copy `best.onnx` to the Pi (rename e.g. `version5.onnx`).
2. Run with the **lower confidence** we discussed:
```bash
python3 detect_pi.py --model version5.onnx --size 640 \
  --conf 0.15 --mold-max-intensity 255 --dark-mold-pct 2.0 --dark-mold-thresh 90 \
  --api https://shelvy-backend.vercel.app
```
- **`--conf 0.15`** = the lowered confidence (catches more mold on the dominant bread)
- Model input is already normalized (÷255) inside; the Pi's brightness handling covers high/low cam.

**Camera at ~15cm** for reliable detection (your locked setup distance).
